```python
class Hero:
    def __init__(self, name, hp, attack_power):
        self.name = name
        self.hp = hp
        self.attack_power = attack_power

    def info(self):
        print(f"Hero: {self.name} | HP: {self.hp} | Power: {self.attack_power}")

    def serang(self, lawan):
        print(f"{self.name} menyerang {lawan.name}!")
        lawan.diserang(self.attack_power)

    def diserang(self, damage):
        self.hp -= damage
        print(f"{self.name} terkena damage {damage}. Sisa HP: {self.hp}")

# --- Main Program ---
print("--- Pertarungan Dimulai ---")
hero1 = Hero("Helcurt", 1150, 150)
hero2 = Hero("Clint", 1225, 200)

hero1.info()
hero2.info()

hero1.serang(hero2)
hero2.serang(hero1)
```

```python
class Hero:
    def __init__(self, nama, hp_awal, attack_power):
        self.nama = nama
        self.__hp = hp_awal # Enkapsulasi: Private attribute
        self.attack_power = attack_power

    def get_hp(self):
        return self.__hp

    def set_hp(self, nilai_baru):
        if nilai_baru < 0:
            self.__hp = 0
        elif nilai_baru > 1000:
            print("Cheat terdeteksi! HP dimaksimalkan ke 1000 saja.")
            self.__hp = 1000
        else:
            self.__hp = nilai_baru

    def diserang(self, damage):
        sisa_hp = self.get_hp() - damage
        self.set_hp(sisa_hp)
        print(f"{self.nama} terkena damage {damage}. Sisa HP: {self.get_hp()}")

    # Add the serang method back to the Hero class
    def serang(self, lawan):
        print(f"{self.nama} menyerang {lawan.nama}!")
        lawan.diserang(self.attack_power)

class Mage(Hero):
    def __init__(self, name, hp, attack_power, mana):
        super().__init__(name, hp, attack_power) # Memanggil constructor Parent
        self.mana = mana

    def skill_fireball(self, lawan):
        if self.mana >= 20:
            print(f"{self.nama} menggunakan Fireball ke {lawan.nama}!")
            self.mana -= 20
            lawan.diserang(self.attack_power * 2)
        else:
            print(f"{self.nama} gagal skill! Mana tidak cukup.")

# --- Main Program ---
hero1 = Hero("Balmond", 200, 10)
eudora = Mage("Eudora", 80, 30, 100)
eudora.skill_fireball(hero1)
```

```python
# Class Mage adalah anak dari class Hero
class Mage(Hero):
  def __init__(self, name, hp, attack_power, mana):
    # Memanggil constructor milik Parent (Hero)
    super().__init__(name, hp, attack_power)
    self.mana = mana

  def info(self):
    print(f"{self.nama} [Mage] | HP: {self.get_hp()} | Mana: {self.mana}") # Corrected HP display

  # Mage punya skill khusus
  def skill_fireball(self, lawan):
    if self.mana >= 20:
      print(f"{self.nama} menggunakan Fireball ke {lawan.nama}!")
      self.mana -= 20
      lawan.diserang(self.attack_power * 2) # Damage 2x lipat
    else:
      print(f"{self.nama} gagal skill! Mana tidak cukup.")

# -- Main Program Baru --
print("\n--- Update Class Hero ---")
eudora = Mage("Eudora", 80, 30, 100)
balmond = Hero("Balmond", 200, 10)

eudora.info()
eudora.serang(balmond) # Serangan biasa (warisan dari Hero)
eudora.skill_fireball(balmond)
```

```python
from abc import ABC, abstractmethod

# 1. Interface / Abstract Class
# Ini adalah KONTRAK. Semua turunan wajib punya method di bawah ini.

class GameUnit(ABC):

  @abstractmethod
  def serang(self, target):
    pass

  @abstractmethod
  def info(self):
    pass

# 2. Implementasi pada Class Konkret
class Hero(GameUnit):
  def __init__(self, nama):
    self.nama = nama

  # Kita WAJIB membuat method serang, kalau tidak akan Error
  def serang(self, target):
    print(f"Hero {self.nama} menebas {target}!")

  def info(self):
    print(f"Saya adalah Hero: {self.nama}")

class Monster(GameUnit):
  def __init__(self, jenis):
    self.jenis = jenis

  # Implementasi serang versi Monster
  def serang(self, target):
    print(f"Monster {self.jenis} menggigit {target}!")

  def info(self):
    print(f"Saya adalah Monster: {self.jenis}")

# -- Uji Coba --
# unit = GameUnit() # ERROR! Abstract class tidak bisa jadi objek.
h = Hero("Alucard")
m = Monster("Serigala")

h.info()
m.info()
```

```python
class Hero:
  def __init__(self, nama):
    self.nama = nama

  def serang(self):
    print("Hero menyerang dengan tangan kosong.")

# Child Class 1
class Mage(Hero):
  def serang(self):
    print(f"{self.nama} (Mage) menembakkan Bola Api! Boom!")

# Child Class 2
class Archer(Hero):
  def serang(self):
    print(f"{self.nama} (Archer) memanah dari jauh! Jleb!")

# Child Class 3
class Fighter(Hero):
  def serang(self):
    print(f"{self.nama} (Fighter) memukul dengan pedang! Slash!")

# Child tambahan
class Healer(Hero):
  def serang(self):
    print(f"{self.nama} (Healer) tidak menyerang, tapi menyembuhkan teman!")

# -- Penerapan Polymorphism --
# Kita punya daftar hero campuran
pasukan = [
  Mage("Eudora"),
  Archer("Miya"),
  Fighter("Zilong"),
  Mage("Gord"),
  Healer("Estes")
]
print("--- PERANG DIMULAI ---")

# Satu perintah loop, tapi respon berbeda-beda (Polymorphism)
for pahlawan in pasukan:
  pahlawan.serang()
```

```python
from abc import ABC, abstractmethod

# Helper function untuk format Rupiah
def format_rp(angka):
    return f"Rp {int(angka):,.0f}".replace(",", ".")

# 1. Abstraction & Encapsulation
class KamarHotel(ABC):
    def __init__(self, nama_kamar, harga_dasar, stok=0):
        self.nama_kamar = nama_kamar
        self.__harga_dasar = harga_dasar
        self.__stok = stok

    @property
    def stok(self):
        return self.__stok

    @property
    def harga_dasar(self):
        return self.__harga_dasar

    def tambah_stok(self, jumlah):
        if jumlah < 0:
            print(f"Gagal update stok {self.jenis_kamar}! Stok tidak boleh negatif ({jumlah}).")
        else:
            self.__stok += jumlah
            print(f"Berhasil menambahkan stok {self.jenis_kamar}: {self.__stok} unit.")

    @abstractmethod
    def tampilkan_detail(self, jumlah_malam, nomor):
        pass

    @abstractmethod
    def hitung_harga_total(self, jumlah_malam):
        pass

# 2. Inheritance & Polymorphism (Kamar Deluxe)
class KamarDeluxe(KamarHotel):
    jenis_kamar = "Kamar Deluxe"

    def __init__(self, nama_kamar, harga_dasar, fasilitas, stok=0):
        super().__init__(nama_kamar, harga_dasar, stok)
        self.fasilitas = fasilitas
        self.pajak = 0.10

    def hitung_harga_total(self, jumlah_malam):
        total_per_malam = self.harga_dasar + (self.harga_dasar * self.pajak)
        return total_per_malam * jumlah_malam

    def tampilkan_detail(self, jumlah_malam, nomor):
        pajak_rp = self.harga_dasar * self.pajak
        subtotal = self.hitung_harga_total(jumlah_malam)
        print(f"{nomor}. [DELUXE] {self.nama_kamar} | Fasilitas: {self.fasilitas}")
        print(f"Harga Dasar/Malam: {format_rp(self.harga_dasar)} | Pajak(10%): {format_rp(pajak_rp)}")
        print(f"Menginap: {jumlah_malam} malam | Subtotal: {format_rp(subtotal)}")

# 3. Inheritance & Polymorphism (Kamar Standard)
class KamarStandard(KamarHotel):
    jenis_kamar = "Kamar Standard"

    def __init__(self, nama_kamar, harga_dasar, kapasitas, stok=0):
        super().__init__(nama_kamar, harga_dasar, stok)
        self.kapasitas = kapasitas
        self.pajak = 0.05

    def hitung_harga_total(self, jumlah_malam):
        total_per_malam = self.harga_dasar + (self.harga_dasar * self.pajak)
        return total_per_malam * jumlah_malam

    def tampilkan_detail(self, jumlah_malam, nomor):
        pajak_rp = self.harga_dasar * self.pajak
        subtotal = self.hitung_harga_total(jumlah_malam)
        print(f"{nomor}. [STANDARD] {self.nama_kamar} | Kapasitas: {self.kapasitas}")
        print(f"Harga Dasar/Malam: {format_rp(self.harga_dasar)} | Pajak(5%): {format_rp(pajak_rp)}")
        print(f"Menginap: {jumlah_malam} malam | Subtotal: {format_rp(subtotal)}")

# Fitur Pemesanan (Di luar class)
def proses_transaksi(daftar_pesanan):
    print("--- STRUK PEMESANAN ---")
    total_tagihan = 0
    for i, pesanan in enumerate(daftar_pesanan, start=1):
        kamar = pesanan['kamar']
        malam = pesanan['malam']

        kamar.tampilkan_detail(malam, i)
        total_tagihan += kamar.hitung_harga_total(malam)

    print("-" * 40)
    print(f"TOTAL TAGIHAN: {format_rp(total_tagihan)}")
    print("-" * 40)

# ==========================================
# Alur Program (User Story)
# ==========================================
if __name__ == "__main__":
    # a) Admin membuat data kamar
    deluxe = KamarDeluxe("Kamar Deluxe Sea View", 1500000, "Private Pool")
    standard = KamarStandard("Kamar Standard Superior", 500000, "2 Orang")

    print("--- SETUP DATA KAMAR ---")
    deluxe.tambah_stok(10)

    # b) Admin mencoba mengisi stok dengan angka negatif
    standard.tambah_stok(-5)

    # Admin mengkoreksi penambahan stok
    standard.tambah_stok(20)

    # c) Tamu memesan kamar
    daftar_pesanan = [
        {'kamar': deluxe, 'malam': 2},
        {'kamar': standard, 'malam': 1}
    ]

    # d) Program menampilkan detail dan tagihan
    proses_transaksi(daftar_pesanan)
```
